```python
# prompt: write me a funtinn that print the percentage of non numeric value in each column and nan valu in each funtion

import numpy as np
def analyze_dataframe(df):
  """
  Analyzes a pandas DataFrame to print the percentage of non-numeric values
  and NaN values in each column.

  Args:
      df: The pandas DataFrame to analyze.
  """

  # Calculate the percentage of non-numeric values for each column
  non_numeric_percentage = (df.applymap(lambda x: not isinstance(x, (int, float, np.number))).sum() / len(df)) * 100
  print("Percentage of Non-Numeric Values per Column:")
  print(non_numeric_percentage)

  # Calculate and print the percentage of NaN values in each column
  print("\nPercentage of NaN Values per Column:")
  for column in df.columns:
    nan_percentage = (df[column].isnull().sum() / len(df)) * 100
    print(f"Column '{column}': {nan_percentage:.2f}% NaN values")

# Example usage with the 'bed' DataFrame
print("Analyzing 'bed' DataFrame:")
analyze_dataframe(bed)
print("\nAnalyzing 'ene' DataFrame:")
analyze_dataframe(ene)
print("\nAnalyzing 'ind' DataFrame:")
analyze_dataframe(ind)
print("\nAnalyzing 'out' DataFrame:")
analyze_dataframe(out)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# def plot_separate_density_and_scatter(df):
#     """
#     Plots individual density graphs for each numeric column and individual
#     scatter plots for each pair of numeric columns in the DataFrame.

#     Parameters:
#     df (pd.DataFrame): Input DataFrame containing numeric columns.

#     Returns:
#     None
#     """
#     # Select only numeric columns
#     numeric_cols = df.select_dtypes(include=['number']).columns.tolist()

#     if not numeric_cols:
#         print("No numeric columns found in the DataFrame.")
#         return

#     # ---- Plot density graph for each column ----
#     for col in numeric_cols:
#         plt.figure(figsize=(8, 5))
#         sns.kdeplot(df[col], fill=True, color='blue')
#         plt.title(f"Density Plot for {col}")
#         plt.xlabel(col)
#         plt.ylabel("Density")
#         plt.grid(True)
#         plt.show()

#     # ---- Plot scatter plot for each unique pair of columns ----
#     for i in range(len(numeric_cols)):
#         for j in range(i + 1, len(numeric_cols)):
#             plt.figure(figsize=(8, 5))
#             sns.scatterplot(x=df[numeric_cols[i]], y=df[numeric_cols[j]], alpha=0.6)
#             plt.title(f"Scatter Plot: {numeric_cols[i]} vs {numeric_cols[j]}")
#             plt.xlabel(numeric_cols[i])
#             plt.ylabel(numeric_cols[j])
#             plt.grid(True)
#             plt.show()



import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_separate_density_and_scatter(df):
    """
    For each numeric column in the DataFrame, this function plots:
    - A density plot to show the distribution.
    - A scatter plot (value vs. index) to help identify outliers.

    Parameters:
    df (pd.DataFrame): The input DataFrame.

    Returns:
    None
    """
    # Select only numeric columns
    numeric_cols = df.select_dtypes(include=['number']).columns.tolist()

    if not numeric_cols:
        print("No numeric columns found in the DataFrame.")
        return

    for col in numeric_cols:
        # Density plot for the column
        plt.figure(figsize=(8, 5))
        sns.kdeplot(df[col], fill=True, color='red')
        plt.title(f"Density Plot for {col}")
        plt.xlabel(col)
        plt.ylabel("Density")
        plt.grid(True)
        plt.show()

        # Scatter plot for the column vs. index (to spot outliers)
        plt.figure(figsize=(8, 5))
        plt.scatter(df.index, df[col], alpha=0.6, color='green')
        plt.title(f"Scatter Plot for {col} (Outlier Detection)")
        plt.xlabel("Index")
        plt.ylabel(col)
        plt.grid(True)
        plt.show()

# Example usage:
# plot_density_and_scatter_outliers(your_dataframe)

# Example usage:
# plot_separate_density_and_scatter(your_dataframe)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_boxplots(df):
    """
    Plots individual boxplots for each numeric column in the DataFrame.
    
    Parameters:
    df (pd.DataFrame): The input DataFrame containing numeric columns.
    
    Returns:
    None
    """
    # Select only numeric columns
    numeric_cols = df.select_dtypes(include=['number']).columns.tolist()

    if not numeric_cols:
        print("No numeric columns found in the DataFrame.")
        return

    for col in numeric_cols:
        plt.figure(figsize=(8, 5))
        sns.boxplot(y=df[col], color='blue', width=0.5)
        plt.title(f"Boxplot for {col}")
        plt.ylabel(col)
        plt.grid(True, linestyle="--", alpha=0.7)
        plt.show()

# Example usage:
# plot_boxplots(your_dataframe)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import zscore

def remove_outliers_zscore(df, threshold=3):
    """
    Removes rows from the DataFrame that contain outliers in any numeric column based on the Z-score method.

    Parameters:
    -----------
    df : pd.DataFrame
        Input DataFrame.
    threshold : float, optional (default=3)
        The Z-score threshold to use. Rows with any numeric column having an absolute z-score above this
        value will be considered outliers and removed.

    Returns:
    --------
    pd.DataFrame
        DataFrame with outliers removed.
    """
    # Only consider numeric columns for outlier removal
    numeric_cols = df.select_dtypes(include=['number']).columns

    # Compute the z-scores for the numeric columns
    z_scores = df[numeric_cols].apply(zscore)

    # Create a boolean mask where True means the row has no outlier in any numeric column
    mask = (np.abs(z_scores) < threshold).all(axis=1)

    # Filter the DataFrame using the mask
    df_clean = df[mask].copy()
    return df_clean

# Example usage:
# cleaned_df = remove_outliers_zscore(your_dataframe, threshold=3)
# print(cleaned_df.shape)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

def decompose_time_series(df, date_col, value_col, model='additive', freq=None):
    """
    Decompose a time series into trend, seasonal, and residual components.

    Parameters
    ----------
    df : pd.DataFrame
        The DataFrame containing the time series data.
    date_col : str
        The name of the column containing datetime information.
    value_col : str
        The name of the column containing the time series values to decompose.
    model : str, optional
        Decomposition model, either 'additive' or 'multiplicative'.
        Default is 'additive'.
    freq : int, optional
        The period (frequency) of the time series. If None,
        the function will attempt to infer the frequency automatically.
        If inference fails, you'll need to provide a value.

    Returns
    -------
    decomposition : DecomposeResult or None
        The statsmodels decomposition result object, containing:
          - trend
          - seasonal
          - resid
          - observed
        Returns None if frequency cannot be inferred.
    """
    # 1. Copy the DataFrame to avoid modifying the original
    data = df.copy()

    # 2. Convert date_col to datetime if not already
    if not pd.api.types.is_datetime64_any_dtype(data[date_col]):
        data[date_col] = pd.to_datetime(data[date_col], errors='coerce')

    # 3. Set date_col as the index and sort by it
    data.set_index(date_col, inplace=True)
    data.sort_index(inplace=True)

    # 4. Drop rows with missing values in the target column
    data.dropna(subset=[value_col], inplace=True)

    # 5. If freq is not provided, try to infer it
    if freq is None:
        inferred = pd.infer_freq(data.index)
        if inferred is None:
            print("Could not infer frequency. Please specify 'freq' manually.")
            return None  # Stop if frequency cannot be inferred
        else:
            # Map common Pandas frequencies to an integer period
            freq_map = {
                'H': 24,   # daily seasonality for hourly data
                'D': 7,    # weekly seasonality for daily data
                'W': 52,   # yearly seasonality for weekly data
                'M': 12,   # yearly seasonality for monthly data
                'Q': 4,    # yearly seasonality for quarterly data
                'A': 1     # yearly seasonality for annual data
            }
            # e.g., inferred could be 'D', 'H', 'W', etc.
            base = inferred[0]  # first character (e.g. 'D')
            freq = freq_map.get(base, 7)  # default to 7 if not found in freq_map

    # 6. Perform the seasonal decomposition
    decomposition = seasonal_decompose(
        data[value_col],
        model=model,
        period=freq
    )

    # 7. Plot the decomposed components
    fig, axes = plt.subplots(4, 1, figsize=(12, 10), sharex=True)

    # Observed / Original
    axes[0].plot(decomposition.observed, label='Observed', color='blue')
    axes[0].set_title('Original Time Series')
    axes[0].legend()

    # Trend
    axes[1].plot(decomposition.trend, label='Trend', color='green')
    axes[1].set_title('Trend Component')
    axes[1].legend()

    # Seasonal
    axes[2].plot(decomposition.seasonal, label='Seasonality', color='orange')
    axes[2].set_title('Seasonal Component')
    axes[2].legend()

    # Residual
    axes[3].plot(decomposition.resid, label='Residuals', color='red')
    axes[3].set_title('Residual Component')
    axes[3].legend()

    plt.tight_layout()
    plt.show()

    return decomposition
```

```python


# Normalize data
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_selected)

# Apply K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_selected['KMeans_Cluster'] = kmeans.fit_predict(df_scaled)

# Apply DBSCAN Clustering
dbscan = DBSCAN(eps=0.5, min_samples=5)
df_selected['DBSCAN_Cluster'] = dbscan.fit_predict(df_scaled)
```

```python

# Compare occupancy clusters with actual working hours
if 'WorkingHours' in df.columns:
    df_selected['Actual_Occupancy'] = df['WorkingHours']
    
```

```python

# Compare Occupancy vs. Actual Energy Consumption Over Time
plt.figure(figsize=(12, 6))
sns.lineplot(x=df['Datetime'], y=df['total_energy'], label="Actual Energy Use", color="blue")
sns.scatterplot(x=df['Datetime'], y=df_selected['KMeans_Cluster'] * df['total_energy'].max() / 2, label="K-Means Clusters", color="red", marker="x")
sns.scatterplot(x=df['Datetime'], y=df_selected['DBSCAN_Cluster'] * df['total_energy'].max() / 2, label="DBSCAN Clusters", color="green", marker="o")
plt.title('Comparison of Occupancy Clusters with Energy Consumption Over Time')
plt.xlabel('Time')
plt.ylabel('Energy Use')
plt.legend()
plt.xticks(rotation=45)
plt.show()

# Silhouette Score for K-Means
print("K-Means Silhouette Score:", silhouette_score(df_scaled, df_selected['KMeans_Cluster']))
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Normalize data before clustering
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_selected[['testo 160 IAQ_51616142 [ppm]']])

# Apply K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_selected['KMeans_Cluster'] = kmeans.fit_predict(df_scaled)

# Extract CO₂ concentration
co2_values = df_selected['testo 160 IAQ_51616142 [ppm]']

# 🔹 **Reassign cluster labels based on CO₂ thresholds**
high_threshold = co2_values.quantile(0.75)  # Upper 25% values considered "High CO₂"
low_threshold = co2_values.quantile(0.25)   # Lower 25% values considered "Low CO₂"

df_selected['CO2_Cluster_Label'] = np.select(
    [
        co2_values >= high_threshold, 
        co2_values < low_threshold
    ], 
    ['High CO2', 'Low CO2'], 
    default='Mid CO2'
)

# 🔹 **CO₂ Clustering Visualization with Corrected Labels**
plt.figure(figsize=(12, 6))
sns.scatterplot(
    x=df['Datetime'], 
    y=co2_values, 
    hue=df_selected['CO2_Cluster_Label'], 
    palette={'Low CO2': 'blue', 'Mid CO2': 'gray', 'High CO2': 'red'}
)
plt.title('Clustering of CO₂ Concentration Using K-Means')
plt.xlabel('Time')
plt.ylabel('CO₂ Concentration [ppm]')
plt.xticks(rotation=45)
plt.legend(title="CO₂ Levels")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score


# Identify the correct timestamp column
print("Energy Data Columns:", ene.columns)
print("Indoor Air Quality Columns:", ind.columns)
print("Outdoor Air Quality Columns:", out.columns)

# Set correct timestamp column (modify if different)
timestamp_col = "Datetime"  # Change this based on print output

# Ensure timestamp column is in all datasets
for df in [ene, ind, out]:
    if timestamp_col not in df.columns:
        raise KeyError(f"Column '{timestamp_col}' not found in {df}")

# Merge datasets on timestamp
df = ene.merge(ind, on=timestamp_col).merge(out, on=timestamp_col)

# Selecting relevant features for clustering
features = ['testo 160 IAQ_51616142 [ppm]', 'testo 160 IAQ_51616142 [°C]', 'total_energy']  # Modify based on actual column names
df_selected = df[[timestamp_col] + features].copy()

# Handle missing values
df_selected.dropna(inplace=True)

# Normalize data
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_selected[features])

# Apply K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_selected['KMeans_Cluster'] = kmeans.fit_predict(df_scaled)

# Apply DBSCAN Clustering
dbscan = DBSCAN(eps=0.5, min_samples=5)  # Tune parameters if needed
df_selected['DBSCAN_Cluster'] = dbscan.fit_predict(df_scaled)

# Add actual working hours (modify column name if different)
working_hours_col = "WorkingHours"  # Change based on dataset
if working_hours_col in df.columns:
    df_selected['Actual_Occupancy'] = df[working_hours_col]
else:
    print(f"Column '{working_hours_col}' not found, skipping actual occupancy comparison.")

# Convert timestamp to datetime for plotting
df_selected[timestamp_col] = pd.to_datetime(df_selected[timestamp_col])

# Visualization: Cluster Distributions Over Time
plt.figure(figsize=(12, 6))
sns.scatterplot(x=df_selected[timestamp_col], y=df_selected['KMeans_Cluster'], hue=df_selected['KMeans_Cluster'], palette="viridis", s=20)
plt.title("K-Means Clustering: Occupancy Distribution Over Time")
plt.xlabel("Time")
plt.ylabel("Cluster ID")
plt.xticks(rotation=45)
plt.show()

# Visualization: Identified Occupancy vs. Actual Working Hours
if working_hours_col in df.columns:
    plt.figure(figsize=(12, 6))
    sns.lineplot(x=df_selected[timestamp_col], y=df_selected['Actual_Occupancy'], label="Actual Working Hours", color="blue")
    sns.scatterplot(x=df_selected[timestamp_col], y=df_selected['KMeans_Cluster'] * df['total_energy'].max() / 2, label="K-Means Clusters", color="red", marker="x")
    plt.title('Comparison of Identified Occupancy vs. Actual Working Hours')
    plt.xlabel('Time')
    plt.ylabel('Occupancy State')
    plt.legend()
    plt.show()

# Visualization: Occupancy vs. Energy Consumption
plt.figure(figsize=(12, 6))
sns.lineplot(x=df_selected[timestamp_col], y=df['total_energy'], label="Actual Energy Use", color="blue")
sns.scatterplot(x=df_selected[timestamp_col], y=df_selected['KMeans_Cluster'] * df['EnergyUse'].max() / 2, label="K-Means Clusters", color="red", marker="x")
plt.title('Comparison of Occupancy Clusters with Energy Consumption')
plt.xlabel('Time')
plt.ylabel('Energy Use')
plt.legend()
plt.xticks(rotation=45)
plt.show()

# Silhouette Score for K-Means
print("K-Means Silhouette Score:", silhouette_score(df_scaled, df_selected['KMeans_Cluster']))
```
